```python
import yfinance as yf

from funcs.plot import trend_diff
```

```python
list_code = [
    "ANT",
    "INTC",
    "AMD",
    "ARM",
    "SFTBY",
    "KXIAY",
    "MU",
    "TSM",
    "NVDA",
    "SKHY",
    "CBRS",
    "CRSP",
    "FLY",
    "LUNR",
    "SPCX",
    "688825.SS",  # CXMT in Shanghai
    "688836.SS",  # Yushu Technology Co, Ltd
]
for code in list_code:
    symbol = f"{code}"
    yticker = yf.Ticker(symbol)
    df = yticker.history(period="3mo", interval="1d")
    df0 = yticker.history(period="6mo", interval="1d")
    print("(", len(df), len(df0), ")")

    if len(df) > 0:
        df["Diff"] = df["High"] - df["Low"]

    if "shortName" in yticker.info:
        name = yticker.info["shortName"]
    elif "longName" in yticker.info:
        name = yticker.info["longName"]
    else:
        name = symbol

    trend_diff(code, df, df0, name)
```
